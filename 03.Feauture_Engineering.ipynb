```python
import pandas as pd 
import numpy as np
from pathlib import Path
files =list(Path.cwd().iterdir())
for file in files:
    print(file.name)
```

```python
gea = pd.read_excel("Gea_Preprocessed_All_Languages.xlsx")
pelagos = pd.read_excel("Pelagos_Preprocessed_All_Languages.xlsx")

print("Gea:", gea.shape)
print("Pelagos:", pelagos.shape)

print("\nGea columns:")
print(gea.columns.tolist())

print("\nPelagos columns:")
print(pelagos.columns.tolist())
```

```python
#Creation of Satisfaction categories
def create_satisfaction(rating):
    if pd.isna(rating):
        return np.nan
    elif rating <= 2.5:
        return "Dissatisfied"
    elif rating <= 3.5:
        return "Neutral/Mixed"
    else:
        return "Satisfied"

gea["Satisfaction_Category"] = gea["Rating"].apply(create_satisfaction)
pelagos["Satisfaction_Category"] = pelagos["Rating"].apply(create_satisfaction)

print("GEA:")
print(gea["Satisfaction_Category"].value_counts(dropna=False))

print("\nPELAGOS:")
print(pelagos["Satisfaction_Category"].value_counts(dropna=False))
```

```python
#Feature region 
def assign_region(country):

    europe = [
        "United Kingdom", "France", "Germany", "Italy", "Portugal",
        "Spain", "Switzerland", "Ireland", "Austria", "Romania",
        "Belgium", "Slovakia", "Poland", "Greece", "Luxembourg"
    ]

    north_america = [
         "United States", "Canada", "Mexico",
         "Dominican Republic", "British Virgin Islands"
    ]

    south_america = [
         "Brazil", "Argentina", "Colombia", "Uruguay"
    ]

    asia = [
         "Japan", "South Korea", "China", "Hong Kong",
         "India", "Singapore", "Malaysia", "Vietnam",
         "Thailand", "Sri Lanka"
    ]

    middle_east = [
         "United Arab Emirates", "Israel", "Turkey",
         "Iraq", "Jordan"
    ]

    oceania = [
        "Australia", "New Zealand"
    ]

    africa = [
        "Tanzania"
    ]

    if country in europe:
        return "Europe"
    elif country in north_america:
        return "North America"
    elif country in south_america:
        return "South America"
    elif country in asia:
        return "Asia"
    elif country in middle_east:
        return "Middle East"
    elif country in oceania:
        return "Oceania"
    elif country in africa:
        return "Africa"
    else:
        return "Unknown"


gea["Guest_Region"] = gea["Guest_Country"].apply(assign_region)
pelagos["Guest_Region"] = pelagos["Guest_Country"].apply(assign_region)

print("GEA:")
print(gea["Guest_Region"].value_counts())

print("\nPELAGOS:")
print(pelagos["Guest_Region"].value_counts())
```
