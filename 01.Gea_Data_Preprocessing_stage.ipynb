```python
import pandas as pd

gea= pd.read_excel("Gea_Reviews_2022_2026_Combined.xlsx",
                  header=1)

print("Total Gea reviews:", len(gea))
print("\nColumns:")
print(gea.columns.tolist())

gea.head()
```

```python
# Create Review_Text

gea["Review Title"] = gea["Review Title"].fillna("").astype(str)
gea["Review"] = gea["Review"].fillna("").astype(str)

gea["Review_Text"] = (
gea["Review Title"].str.strip() + " " +
gea["Review"].str.strip()
).str.strip()

print("Total reviews:", len(gea))
print("Missing Review_Text:", gea["Review_Text"].eq("").sum())

gea[["Review Site", "Review Title", "Review", "Review_Text"]].head(10)
```

```python
# Remove rows with no review text

gea = gea[gea["Review_Text"].str.strip() != ""].copy()
gea.reset_index(drop=True, inplace=True)

print("Reviews remaining:", len(gea))
```

```python
print("Language values:")
print(gea["Language"].value_counts(dropna=False))
```

```python
gea[gea["Language"]== "Language"][
    ["Review Site", "Review_Text", "Language"]
    ]
```

```python
#Remove repeated header rows as they are not reviews 
gea = gea[gea["Language"] != "Language"].copy()
gea.reset_index(drop=True, inplace=True)

print("Reviews remaining:", len (gea))
```

```python
print(gea["Language"].value_counts())
```

```python
# Remove repeated header rows

gea = gea[gea["Language"] != "Language"].copy()
gea.reset_index(drop=True, inplace=True)

print("Reviews remaining:", len(gea))
print(gea["Language"].value_counts())
```

```python
# Check duplicate reviews

print("Total reviews:", len(gea))
print("Duplicate reviews:", gea["Review_Text"].duplicated().sum())
```

```python
#Removing duplicate reviews 
gea = gea.drop_duplicates(
    subset=["Review_Text"],
    keep="first").copy()

gea.reset_index(drop=True, inplace=True)
print("Reviews after removing duplicates:", len(gea))
```

```python
#Checking short reviews
gea["Review_Length"] = gea["Review_Text"].astype(str).str.strip().str.len()
gea["Short_Review"] = gea["Review_Length"] <10
```

```python
print("Total reviews:", len(gea))
print("Short reviews:", gea["Short_Review"].sum())
```

```python
#Creating English dataset for NLP
gea_english = gea[
    gea["Language"] == "English"
    ].copy()
print("All reviews:", len(gea))
print("English reviews:", len(gea_english))
print("Other languages:", len(gea) - len(gea_english))
```

```python
import re

gea_english["Clean_Text"] = (
gea_english["Review_Text"]
.astype(str)
.str.lower()
.str.replace(r"http\S+|www\S+", "", regex=True)
.str.replace(r"[^a-z\s]", " ", regex=True)
.str.replace(r"\s+", " ", regex=True)
.str.strip()
)

gea_english[["Review_Text", "Clean_Text"]].head(10)
```

```python
import nltk
nltk.download("stopwords")
nltk.download("wordnet")

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

gea_english["Processed_Text"] = gea_english["Clean_Text"].apply(
lambda text: " ".join(
lemmatizer.lemmatize(word)
for word in text.split()
if word not in stop_words
)
)

gea_english[["Clean_Text", "Processed_Text"]].head(10)
```

```python
gea.to_excel("Gea_Preprocessed_All_Languages.xlsx", index=False)

gea_english.to_excel("Gea_Preprocessed_English.xlsx", index=False)

print("Files saved!")
```
