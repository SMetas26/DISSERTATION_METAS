```python
# Create combined text field for NLP analysis

gea["Review_Text"] = (
gea["Review Title"].fillna("").astype(str).str.strip()
+ " "
+ gea["Review"].fillna("").astype(str).str.strip()
).str.strip()

pelagos["Review_Text"] = (
pelagos["Review Title"].fillna("").astype(str).str.strip()
+ " "
+ pelagos["Review"].fillna("").astype(str).str.strip()
).str.strip()

# Count reviews containing usable text
print("Gea reviews with text:", (gea["Review_Text"] != "").sum())
print("Pelagos reviews with text:", (pelagos["Review_Text"] != "").sum())

# Preview
gea[["Review Title", "Review", "Review_Text"]].head()
```

```python
#Creation of NLP dataset containing only reviews with usable text
gea_nlp = gea[gea["Review_Text"].str.strip() != ""].copy()
pelagos_nlp = pelagos[pelagos["Review_Text"].str.strip() != ""].copy()

#To reset the index
gea_nlp.reset_index(drop=True, inplace=True)
pelagos_nlp.reset_index(drop=True, inplace=True)

print("Original Gea:", len(gea))
print("Gea NLP dataset:", len(gea_nlp))

print("Original Pelagos:", len(pelagos))
print("Pelagos NLP datast:", len(pelagos_nlp))

print("Total reviews available for NLP:",
      len(gea_nlp) + len(pelagos_nlp))
```

```python
#Checking language distribution in NLP datasets

print("Gea - Languages")
print(gea_nlp["Language"].value_counts(dropna=False))

print("\n" + "=" * 50)

print("Pelagos - Languages")
print(pelagos_nlp["Language"].value_counts(dropna=False))
```

```python
from langdetect import detect, DetectorFactory

DetectorFactory.seed = 0

def detect_language(text):
    try:
        if pd.isna(text) or len(str(text).strip()) < 3:
            return "unknown"
        return detect(str(text))
    except:
        return "unknown"

    pelagos_unknown_language= pelagos_nlp[pelagos_nlp["Language"].isna()
        ].copy()

pelagos_unknown_language["Detect_Language"]= (pelagos_unknown_language["Review_Text"].apply(detect_language)
                                             )

pelagos_unknown_language["Detect_Language"].value_counts(dropna=False)
```

```python
pelagos_unknown_language[
    pelagos_unknown_language["Detect_Language"] !="el"
][["Review Site", "Review_Text", "Detect_Language"]]
```

```python
#Creating a final language column while preserving the original labels from Revinate 
pelagos_nlp["Language_Final"] = pelagos_nlp["Language"]

# missing language reviews are confidently detected as Greek
greek_indices =pelagos_unknown_language[
    pelagos_unknown_language["Detect_Language"] == "el"].index
pelagos_nlp.loc[greek_indices, "Language_Final"] = "Greek"

# the 3 ambiguous short reviews to be labeled as Other or Unknown
other_indices= pelagos_unknown_language[
    pelagos_unknown_language["Detect_Language"] != "el"
    ].index
pelagos_nlp.loc[other_indices, "Language_Final"] = "Other/Unknown"

pelagos_nlp["Language_Final"].value_counts(dropna=False)
```

```python
#Check for missing or short review text 
pelagos_nlp["Review_Length"] = (
pelagos_nlp["Review_Text"]
.fillna("")
.astype(str)
.str.strip()
.str.len()
)

print("Total reviews:", len(pelagos_nlp))
print("Missing Review_Text:", pelagos_nlp["Review_Text"].isna().sum())
print("Empty reviews:", (pelagos_nlp["Review_Length"] == 0).sum())
print("Very short reviews (under 10 characters):",
(pelagos_nlp["Review_Length"] < 10).sum())
```

```python
#Inspecting the very short reviews before deciding the keep them or not 
pelagos_nlp[
    pelagos_nlp["Review_Length"] <10
    ][["Review_Text", "Language_Final", "Review_Length"]]
```

```python
import re

pelagos_nlp["Clean_Text"] = (
pelagos_nlp["Review_Text"]
.astype(str)
.str.lower()
.str.replace(r"http\S+|www\S+", " ", regex=True)
.str.replace(r"\s+", " ", regex=True)
.str.strip()
)

pelagos_nlp[["Review_Text", "Clean_Text"]].head(10)
```

```python
# Checking the data before continuing

print("Total reviews:", len(pelagos_nlp))

print("Missing reviews:",
pelagos_nlp["Review_Text"].isna().sum())

print("Short reviews:",
(pelagos_nlp["Review_Length"] < 10).sum())

print("Duplicate reviews:",
pelagos_nlp["Review_Text"].duplicated().sum())

print("\nLanguages:")
print(pelagos_nlp["Language_Final"].value_counts())
```

```python
# Check repeated review texts

duplicates = pelagos_nlp[
pelagos_nlp.duplicated(subset=["Review_Text"], keep=False)
]

duplicates[
["Review_Text", "Language_Final", "Review_Length"]
].sort_values("Review_Text").head(30)
```

```python
# Remove exact duplicate reviews

pelagos_nlp = pelagos_nlp.drop_duplicates(
subset=["Review_Text"],
keep="first"
).copy()

pelagos_nlp.reset_index(drop=True, inplace=True)

print("Reviews after removing duplicates:", len(pelagos_nlp))
```

```python
# Mark short reviews

pelagos_nlp["Short_Review"] = pelagos_nlp["Review_Length"] < 10

print("Total reviews:", len(pelagos_nlp))
print("Short reviews:", pelagos_nlp["Short_Review"].sum())
```

```python
# Create English dataset for NLP

pelagos_english = pelagos_nlp[
pelagos_nlp["Language_Final"] == "English"
].copy()

print("All reviews:", len(pelagos_nlp))
print("English reviews:", len(pelagos_english))
print("Other languages:", len(pelagos_nlp) - len(pelagos_english))
```

```python
import nltk
nltk.download("stopwords")
nltk.download("wordnet")

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

pelagos_english["Processed_Text"] = pelagos_english["Clean_Text"].apply(
lambda text: " ".join(
lemmatizer.lemmatize(word)
for word in text.split()
if word.isalpha() and word not in stop_words
)
)

pelagos_english[["Clean_Text", "Processed_Text"]].head(10)
```
